# Exploring London's Travel Network (SQL)

**Goal:** analyze public transport usage in London using Transport for London (TfL) open data, to understand which modes of transport are used most and how demand changed over time.

**Tools:** SQL (Snowflake), run in DataCamp DataLab.
**Data:** `TFL.JOURNEYS` – monthly journeys (in millions) by transport type, 2010–2022. A copy of the data is included as `tfl_journeys_final.csv`.

| Column | Description |
|---|---|
| `MONTH` / `YEAR` | Period of the record |
| `DAYS` | Number of days in the period |
| `REPORT_DATE` | Date the data was reported |
| `JOURNEY_TYPE` | Mode of transport |
| `JOURNEYS_MILLIONS` | Journeys made, in millions |

## 1. Which transport types are used the most?
Total journeys per transport type across the whole period.

```python
SELECT JOURNEY_TYPE,
       SUM(JOURNEYS_MILLIONS) AS TOTAL_JOURNEYS_MILLIONS
FROM TFL.JOURNEYS
GROUP BY JOURNEY_TYPE
ORDER BY TOTAL_JOURNEYS_MILLIONS DESC;
```

## 2. When was the Emirates Airline cable car most popular?
Top 5 months by number of journeys (rounded to two decimals).

```python
SELECT MONTH,
       YEAR,
       ROUND(JOURNEYS_MILLIONS, 2) AS ROUNDED_JOURNEYS_MILLIONS
FROM TFL.JOURNEYS
WHERE JOURNEY_TYPE = 'Emirates Airline'
  AND JOURNEYS_MILLIONS IS NOT NULL
ORDER BY ROUNDED_JOURNEYS_MILLIONS DESC
LIMIT 5;
```

## 3. Which years had the lowest Underground & DLR usage?

```python
SELECT YEAR,
       JOURNEY_TYPE,
       SUM(JOURNEYS_MILLIONS) AS TOTAL_JOURNEYS_MILLIONS
FROM TFL.JOURNEYS
WHERE JOURNEY_TYPE LIKE '%Underground%'
GROUP BY YEAR, JOURNEY_TYPE
ORDER BY TOTAL_JOURNEYS_MILLIONS ASC
LIMIT 5;
```

## Conclusions

- **The bus is London's backbone:** ~24,900 million journeys, about 59% of all journeys in the period and well ahead of the Underground & DLR (~15,000 million).
- **The cable car was a novelty effect:** its three best months are all in 2012, its first year of operation (peak: May 2012, 0.53 million journeys). Demand never got back to those levels.
- **COVID-19 had a huge impact on the Tube:** 2020, 2021 and 2022 are the three weakest years. In 2020 usage fell to 310 million journeys, less than a third of 2010, and by 2022 it had still not recovered to 2010 levels.
