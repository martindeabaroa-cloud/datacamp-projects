# Analyzing International Students' Mental Health (SQL)

**Context:** a 2018 survey at a Japanese international university found that international students have a higher risk of mental health difficulties, and that social connectedness and acculturative stress (stress from adapting to a new culture) are predictors of depression.

**Goal:** check whether **length of stay** is related to depression, social connectedness and acculturative stress among international students.

**Tools:** SQL (PostgreSQL), run in DataCamp DataLab.
**Data:** `students` table (286 students, included as `students.csv`).

| Column | Description |
|---|---|
| `inter_dom` | International or domestic student |
| `stay` | Length of stay, in years |
| `todep` | Depression score (PHQ-9 test) |
| `tosc` | Social connectedness score (SCS test) |
| `toas` | Acculturative stress score (ASISS test) |

## Average scores of international students by length of stay

```python
SELECT stay,
       COUNT(*)              AS count_int,
       ROUND(AVG(todep), 2)  AS average_phq,
       ROUND(AVG(tosc), 2)   AS average_scs,
       ROUND(AVG(toas), 2)   AS average_as
FROM students
WHERE inter_dom = 'Inter'
GROUP BY stay
ORDER BY stay DESC;
```

## Conclusions

- **Most international students have been there 1–3 years** (180 of 201). Groups with 5+ years have only 1–3 students, so their averages are not reliable.
- **Acculturative stress rises during the first years:** from 72.8 (1 year) to 78 (2–3 years) and 87.7 (4 years). It does not go down with time, as one might expect.
- **Depression stays fairly stable** (PHQ-9 ≈ 7.5–9 for 1–4 years), slightly higher at 3 years.
- **Social connectedness decreases slightly** from year 1 (38.1) to year 4 (33.9).
- Overall, length of stay alone does not show a strong effect on depression. A next step would be to analyze the correlation between depression, social connectedness and acculturative stress directly.
